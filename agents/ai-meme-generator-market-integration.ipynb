````python
class MemeFormat(Enum):
    DRAKE_POINTING = "drake_pointing"
    DISTRACTED_BOYFRIEND = "distracted_boyfriend"
    WOMAN_YELLING_CAT = "woman_yelling_cat"
    EXPANDING_BRAIN = "expanding_brain"
    TWO_BUTTONS = "two_buttons"
    SURPRISED_PIKACHU = "surprised_pikachu"
    CHANGE_MY_MIND = "change_my_mind"
    THIS_IS_FINE = "this_is_fine"
    GALAXY_BRAIN = "galaxy_brain"
    STONKS = "stonks"

class Platform(Enum):
    TWITTER = "twitter"
    REDDIT = "reddit"
    INSTAGRAM = "instagram"
    TIKTOK = "tiktok"
    LINKEDIN = "linkedin"
    FACEBOOK = "facebook"

class ViralityLevel(Enum):
    LOW = "low"
    MEDIUM = "medium"
    HIGH = "high"
    VIRAL = "viral"

@dataclass
class TrendingTopic:
    """Represents a trending topic or event"""
    topic: str
    platform: Platform
    trend_score: float
    sentiment: str
    hashtags: List[str]
    keywords: List[str]
    discovered_at: datetime
    related_topics: List[str] = field(default_factory=list)

@dataclass
class MemeTemplate:
    """Represents a meme template with metadata"""
    template_id: str
    name: str
    format: MemeFormat
    text_positions: List[Dict[str, Any]]
    image_path: Optional[str] = None
    popularity_score: float = 0.5
    usage_context: List[str] = field(default_factory=list)
    text_style: Dict[str, Any] = field(default_factory=dict)

@dataclass
class GeneratedMeme:
    """Represents a generated meme with metadata"""
    meme_id: str
    template: MemeTemplate
    text_content: List[str]
    topic: str
    virality_score: float
    sentiment: str
    hashtags: List[str]
    target_platforms: List[Platform]
    brand_safe: bool
    generated_at: datetime = field(default_factory=datetime.now)
    image_data: Optional[bytes] = None
    performance_metrics: Dict[str, Any] = field(default_factory=dict)

@dataclass
class MarketEvent:
    """Represents a market event or cultural moment"""
    event_id: str
    title: str
    description: str
    event_type: str  # "earnings", "product_launch", "cultural", "political", etc.
    impact_score: float
    sentiment: str
    related_stocks: List[str] = field(default_factory=list)
    related_brands: List[str] = field(default_factory=list)
    occurred_at: datetime = field(default_factory=datetime.now)
    meme_potential: float = 0.5
````

````python
class TrendAnalyzer:
    """Analyzes current trends and market events for meme opportunities"""
    
    def __init__(self):
        self.trend_cache = {}
        self.sentiment_analyzer = None
        
        # Initialize sentiment analysis pipeline
        if pipeline:
            try:
                self.sentiment_analyzer = pipeline(
                    "sentiment-analysis",
                    model="cardiffnlp/twitter-roberta-base-sentiment-latest",
                    return_all_scores=True
                )
            except Exception as e:
                logger.warning(f"Could not load sentiment model: {e}")
                self.sentiment_analyzer = None
    
    def get_trending_topics(self, platform: Platform, limit: int = 10) -> List[TrendingTopic]:
        """Get trending topics from various platforms"""
        cache_key = f"{platform.value}_{datetime.now().hour}"
        
        if cache_key in self.trend_cache:
            return self.trend_cache[cache_key]
        
        # Simulate trending topics (in production, this would use real APIs)
        trending_topics = self._simulate_trending_topics(platform, limit)
        
        self.trend_cache[cache_key] = trending_topics
        return trending_topics
    
    def _simulate_trending_topics(self, platform: Platform, limit: int) -> List[TrendingTopic]:
        """Simulate trending topics for demo purposes"""
        sample_topics = {
            Platform.TWITTER: [
                ("AI Revolution", ["#AI", "#Technology", "#Future"], ["artificial intelligence", "machine learning", "automation"]),
                ("Stock Market Rally", ["#Stocks", "#Finance", "#Investing"], ["bull market", "earnings", "portfolio"]),
                ("Climate Change Summit", ["#Climate", "#Environment", "#Sustainability"], ["global warming", "renewable energy", "carbon neutral"]),
                ("New iPhone Launch", ["#iPhone", "#Apple", "#Tech"], ["smartphone", "innovation", "features"]),
                ("Cryptocurrency Surge", ["#Crypto", "#Bitcoin", "#DeFi"], ["blockchain", "digital currency", "investment"])
            ],
            Platform.REDDIT: [
                ("Gaming Industry Drama", ["#Gaming", "#Drama", "#Industry"], ["video games", "controversy", "community"]),
                ("Meme Stock Phenomenon", ["#Stonks", "#WSB", "#Finance"], ["wallstreetbets", "retail investors", "squeeze"]),
                ("Space Exploration News", ["#Space", "#NASA", "#Mars"], ["rocket", "exploration", "astronauts"]),
                ("Tech Layoffs Wave", ["#Tech", "#Jobs", "#Economy"], ["unemployment", "startups", "hiring freeze"]),
                ("Pop Culture Controversy", ["#PopCulture", "#Celebrity", "#Drama"], ["entertainment", "social media", "scandal"])
            ]
        }
        
        platform_topics = sample_topics.get(platform, sample_topics[Platform.TWITTER])
        trending_topics = []
        
        for i, (topic, hashtags, keywords) in enumerate(platform_topics[:limit]):
            sentiment = self._analyze_sentiment(topic + " " + " ".join(keywords))
            
            trending_topic = TrendingTopic(
                topic=topic,
                platform=platform,
                trend_score=random.uniform(0.6, 1.0),
                sentiment=sentiment,
                hashtags=hashtags,
                keywords=keywords,
                discovered_at=datetime.now(),
                related_topics=[]
            )
            
            trending_topics.append(trending_topic)
        
        return trending_topics
    
    def _analyze_sentiment(self, text: str) -> str:
        """Analyze sentiment of text"""
        if self.sentiment_analyzer:
            try:
                result = self.sentiment_analyzer(text)[0]
                # Get the highest scoring sentiment
                best_sentiment = max(result, key=lambda x: x['score'])
                return best_sentiment['label'].lower()
            except Exception as e:
                logger.warning(f"Sentiment analysis failed: {e}")
        
        # Fallback to simple keyword-based sentiment
        positive_words = ['good', 'great', 'amazing', 'awesome', 'excellent', 'success', 'win', 'rise', 'up']
        negative_words = ['bad', 'terrible', 'awful', 'fail', 'crash', 'down', 'loss', 'drop', 'decline']
        
        text_lower = text.lower()
        positive_count = sum(1 for word in positive_words if word in text_lower)
        negative_count = sum(1 for word in negative_words if word in text_lower)
        
        if positive_count > negative_count:
            return 'positive'
        elif negative_count > positive_count:
            return 'negative'
        else:
            return 'neutral'
    
    def detect_market_events(self) -> List[MarketEvent]:
        """Detect significant market events and cultural moments"""
        # In production, this would integrate with financial APIs, news feeds, etc.
        sample_events = [
            MarketEvent(
                event_id="event_1",
                title="Tech Giant Earnings Beat",
                description="Major technology company reports record quarterly earnings",
                event_type="earnings",
                impact_score=0.8,
                sentiment="positive",
                related_stocks=["AAPL", "GOOGL", "MSFT"],
                related_brands=["Apple", "Google", "Microsoft"],
                meme_potential=0.7
            ),
            MarketEvent(
                event_id="event_2",
                title="Viral Social Media Challenge",
                description="New dance challenge takes social media by storm",
                event_type="cultural",
                impact_score=0.6,
                sentiment="positive",
                related_brands=["TikTok", "Instagram"],
                meme_potential=0.9
            ),
            MarketEvent(
                event_id="event_3",
                title="Cryptocurrency Market Volatility",
                description="Major cryptocurrency experiences significant price swings",
                event_type="market",
                impact_score=0.7,
                sentiment="neutral",
                related_stocks=["COIN", "MSTR"],
                related_brands=["Bitcoin", "Ethereum"],
                meme_potential=0.8
            )
        ]
        
        return sample_events
    
    def calculate_meme_opportunity_score(self, topic: TrendingTopic, event: MarketEvent = None) -> float:
        """Calculate the meme opportunity score for a topic/event combination"""
        score = 0.0
        
        # Base score from trend strength
        score += topic.trend_score * 0.3
        
        # Sentiment bonus (controversial/extreme sentiments often generate more memes)
        if topic.sentiment in ['very_positive', 'very_negative']:
            score += 0.2
        elif topic.sentiment in ['positive', 'negative']:
            score += 0.1
        
        # Platform-specific bonuses
        platform_bonuses = {
            Platform.TWITTER: 0.1,
            Platform.REDDIT: 0.15,
            Platform.TIKTOK: 0.1,
            Platform.INSTAGRAM: 0.05
        }
        score += platform_bonuses.get(topic.platform, 0.0)
        
        # Event impact bonus
        if event:
            score += event.meme_potential * 0.2
            score += event.impact_score * 0.1
        
        # Keyword-based bonus
        meme_friendly_keywords = ['fail', 'success', 'drama', 'controversy', 'unexpected', 'viral', 'trending']
        keyword_matches = sum(1 for keyword in topic.keywords if any(meme_word in keyword.lower() for meme_word in meme_friendly_keywords))
        score += keyword_matches * 0.05
        
        return min(1.0, score)  # Cap at 1.0
    
    def get_related_topics(self, main_topic: str, all_topics: List[TrendingTopic]) -> List[str]:
        """Find topics related to the main topic"""
        related = []
        main_keywords = main_topic.lower().split()
        
        for topic in all_topics:
            if topic.topic.lower() != main_topic.lower():
                # Check for keyword overlap
                topic_keywords = topic.topic.lower().split() + [kw.lower() for kw in topic.keywords]
                overlap = len(set(main_keywords) & set(topic_keywords))
                
                if overlap > 0:
                    related.append(topic.topic)
        
        return related[:3]  # Return top 3 related topics
````

````python
class MemeTextGenerator:
    """Generates meme text using transformer models"""
    
    def __init__(self):
        self.text_generator = None
        self.tokenizer = None
        self.model = None
        
        # Initialize text generation pipeline
        if pipeline and torch:
            try:
                self.text_generator = pipeline(
                    "text-generation",
                    model="microsoft/DialoGPT-medium",
                    tokenizer="microsoft/DialoGPT-medium",
                    device=0 if torch.cuda.is_available() else -1
                )
            except Exception as e:
                logger.warning(f"Could not load text generation model: {e}")
                self.text_generator = None
        
        # Load meme text templates
        self.meme_templates = self._load_meme_templates()
    
    def _load_meme_templates(self) -> Dict[MemeFormat, List[str]]:
        """Load meme text templates for different formats"""
        return {
            MemeFormat.DRAKE_POINTING: [
                ["{old_thing}", "{new_thing}"],
                ["{bad_option}", "{good_option}"],
                ["{outdated_approach}", "{modern_approach}"]
            ],
            MemeFormat.DISTRACTED_BOYFRIEND: [
                ["{current_thing}", "{new_attractive_thing}", "{upset_thing}"],
                ["{old_technology}", "{new_technology}", "{legacy_system}"]
            ],
            MemeFormat.WOMAN_YELLING_CAT: [
                ["{angry_statement}", "{confused_response}"],
                ["{accusation}", "{innocent_reaction}"]
            ],
            MemeFormat.EXPANDING_BRAIN: [
                ["{basic_idea}", "{better_idea}", "{great_idea}", "{galaxy_brain_idea}"],
                ["{level_1}", "{level_2}", "{level_3}", "{level_4}"]
            ],
            MemeFormat.TWO_BUTTONS: [
                ["{difficult_choice_1}", "{difficult_choice_2}"],
                ["{option_a}", "{option_b}"]
            ],
            MemeFormat.SURPRISED_PIKACHU: [
                ["{obvious_cause}", "{shocked_reaction_to_obvious_effect}"],
                ["{predictable_action}", "{surprise_at_predictable_consequence}"]
            ],
            MemeFormat.CHANGE_MY_MIND: [
                ["{controversial_opinion}"],
                ["{unpopular_truth}"]
            ],
            MemeFormat.THIS_IS_FINE: [
                ["{disaster_situation}", "This is fine"],
                ["{everything_going_wrong}", "Everything is fine"]
            ],
            MemeFormat.STONKS: [
                ["{investment_decision}", "STONKS"],
                ["{financial_strategy}", "📈"]
            ]
        }
    
    def generate_meme_text(self, topic: TrendingTopic, meme_format: MemeFormat, 
                          context: str = "") -> List[str]:
        """Generate meme text for a specific topic and format"""
        if self.text_generator:
            return self._generate_with_ai(topic, meme_format, context)
        else:
            return self._generate_with_templates(topic, meme_format, context)
    
    def _generate_with_ai(self, topic: TrendingTopic, meme_format: MemeFormat, 
                         context: str) -> List[str]:
        """Generate meme text using AI models"""
        try:
            # Create prompt based on meme format and topic
            prompt = self._create_generation_prompt(topic, meme_format, context)
            
            # Generate text
            generated = self.text_generator(
                prompt,
                max_length=len(prompt.split()) + 20,
                num_return_sequences=3,
                temperature=0.8,
                do_sample=True,
                pad_token_id=50256
            )
            
            # Extract and clean generated text
            meme_texts = []
            for result in generated:
                text = result['generated_text'].replace(prompt, '').strip()
                if text and len(text) < 100:  # Keep it meme-length
                    meme_texts.append(text)
            
            return meme_texts[:2] if meme_texts else self._generate_with_templates(topic, meme_format, context)
            
        except Exception as e:
            logger.warning(f"AI generation failed: {e}")
            return self._generate_with_templates(topic, meme_format, context)
    
    def _generate_with_templates(self, topic: TrendingTopic, meme_format: MemeFormat, 
                                context: str) -> List[str]:
        """Generate meme text using templates"""
        templates = self.meme_templates.get(meme_format, [["{topic}"]])
        template = random.choice(templates)
        
        # Fill template with topic-relevant content
        filled_template = []
        
        for text_slot in template:
            filled_text = self._fill_template_slot(text_slot, topic, context)
            filled_template.append(filled_text)
        
        return filled_template
    
    def _create_generation_prompt(self, topic: TrendingTopic, meme_format: MemeFormat, 
                                 context: str) -> str:
        """Create a prompt for AI text generation"""
        format_descriptions = {
            MemeFormat.DRAKE_POINTING: "Create a comparison meme where Drake rejects one thing and points approvingly at another",
            MemeFormat.DISTRACTED_BOYFRIEND: "Create a meme about being distracted by something new while in a relationship with something else",
            MemeFormat.WOMAN_YELLING_CAT: "Create a meme with an angry accusation and a confused innocent response",
            MemeFormat.EXPANDING_BRAIN: "Create a meme showing progressively more enlightened thoughts",
            MemeFormat.SURPRISED_PIKACHU: "Create a meme about being surprised by an obvious consequence",
            MemeFormat.CHANGE_MY_MIND: "Create a controversial opinion statement",
            MemeFormat.THIS_IS_FINE: "Create a meme about staying calm during chaos",
            MemeFormat.STONKS: "Create a meme about questionable financial decisions"
        }
        
        description = format_descriptions.get(meme_format, "Create a funny meme")
        
        prompt = f"""Create a funny meme about {topic.topic}. {description}. 
Context: {context}
Keywords: {', '.join(topic.keywords)}
Sentiment: {topic.sentiment}
Meme text:"""
        
        return prompt
    
    def _fill_template_slot(self, slot: str, topic: TrendingTopic, context: str) -> str:
        """Fill a template slot with relevant content"""
        # Define context-aware replacements
        replacements = {
            "{topic}": topic.topic,
            "{old_thing}": self._get_old_alternative(topic),
            "{new_thing}": topic.topic,
            "{bad_option}": self._get_negative_alternative(topic),
            "{good_option}": topic.topic,
            "{current_thing}": "Current situation",
            "{new_attractive_thing}": topic.topic,
            "{upset_thing}": "Old way of doing things",
            "{angry_statement}": f"You can't just {topic.keywords[0] if topic.keywords else 'do that'}!",
            "{confused_response}": "😐",
            "{basic_idea}": self._get_basic_version(topic),
            "{better_idea}": topic.topic,
            "{great_idea}": f"Advanced {topic.topic}",
            "{galaxy_brain_idea}": f"Quantum {topic.topic}",
            "{difficult_choice_1}": f"Ignore {topic.topic}",
            "{difficult_choice_2}": f"Embrace {topic.topic}",
            "{obvious_cause}": f"When you {topic.keywords[0] if topic.keywords else 'do something'}",
            "{shocked_reaction_to_obvious_effect}": "*Surprised Pikachu face*",
            "{controversial_opinion}": f"{topic.topic} is overrated",
            "{disaster_situation}": f"Everything about {topic.topic} is broken",
            "{investment_decision}": f"Investing in {topic.topic}"
        }
        
        # Apply replacements
        result = slot
        for placeholder, replacement in replacements.items():
            result = result.replace(placeholder, replacement)
        
        return result
    
    def _get_old_alternative(self, topic: TrendingTopic) -> str:
        """Get an old/outdated alternative to the topic"""
        alternatives = {
            "ai": "Manual processes",
            "crypto": "Traditional banking",
            "remote work": "Office work",
            "streaming": "Cable TV",
            "electric cars": "Gas cars"
        }
        
        topic_lower = topic.topic.lower()
        for key, alt in alternatives.items():
            if key in topic_lower:
                return alt
        
        return "Old way"
    
    def _get_negative_alternative(self, topic: TrendingTopic) -> str:
        """Get a negative alternative to the topic"""
        if topic.sentiment == 'positive':
            return f"Ignoring {topic.topic}"
        else:
            return f"Dealing with {topic.topic}"
    
    def _get_basic_version(self, topic: TrendingTopic) -> str:
        """Get a basic version of the topic"""
        return f"Basic {topic.topic.lower()}"
    
    def optimize_for_platform(self, text: List[str], platform: Platform) -> List[str]:
        """Optimize meme text for specific platform constraints"""
        optimized = []
        
        for t in text:
            if platform == Platform.TWITTER:
                # Twitter has character limits
                if len(t) > 50:
                    t = t[:47] + "..."
            elif platform == Platform.INSTAGRAM:
                # Instagram prefers shorter, punchier text
                if len(t) > 30:
                    t = t[:27] + "..."
            elif platform == Platform.TIKTOK:
                # TikTok prefers very short text
                if len(t) > 25:
                    t = t[:22] + "..."
            
            optimized.append(t)
        
        return optimized
````

````python
class ViralityPredictor:
    """Predicts the viral potential of generated memes"""
    
    def __init__(self):
        self.engagement_patterns = self._load_engagement_patterns()
        self.viral_keywords = self._load_viral_keywords()
        self.classifier = None
        
        # Initialize classification pipeline for content analysis
        if pipeline:
            try:
                self.classifier = pipeline(
                    "text-classification",
                    model="cardiffnlp/twitter-roberta-base-emotion",
                    return_all_scores=True
                )
            except Exception as e:
                logger.warning(f"Could not load emotion classification model: {e}")
                self.classifier = None
    
    def _load_engagement_patterns(self) -> Dict[str, float]:
        """Load historical engagement patterns"""
        return {
            "humor": 0.8,
            "controversy": 0.7,
            "relatability": 0.9,
            "current_events": 0.6,
            "nostalgia": 0.5,
            "meta_humor": 0.7,
            "absurdity": 0.6,
            "social_commentary": 0.5
        }
    
    def _load_viral_keywords(self) -> Dict[str, float]:
        """Load keywords that tend to increase viral potential"""
        return {
            "when": 0.3,
            "me": 0.4,
            "literally": 0.2,
            "nobody": 0.3,
            "everyone": 0.2,
            "meanwhile": 0.2,
            "pov": 0.4,
            "imagine": 0.3,
            "that moment when": 0.4,
            "trying to": 0.2,
            "be like": 0.3,
            "mood": 0.3,
            "facts": 0.2,
            "no cap": 0.3,
            "fr": 0.2,
            "periodt": 0.2
        }
    
    def predict_virality(self, meme: GeneratedMeme, context: Dict[str, Any] = None) -> float:
        """Predict the viral potential of a meme"""
        score = 0.0
        factors = []
        
        # Analyze text content
        text_score = self._analyze_text_virality(meme.text_content)
        score += text_score * 0.4
        factors.append(("text_analysis", text_score))
        
        # Template popularity
        template_score = meme.template.popularity_score
        score += template_score * 0.2
        factors.append(("template_popularity", template_score))
        
        # Topic relevance and trend strength
        topic_score = self._analyze_topic_virality(meme.topic, context)
        score += topic_score * 0.2
        factors.append(("topic_relevance", topic_score))
        
        # Sentiment and emotion analysis
        emotion_score = self._analyze_emotional_impact(meme.text_content)
        score += emotion_score * 0.1
        factors.append(("emotional_impact", emotion_score))
        
        # Platform optimization
        platform_score = self._analyze_platform_fit(meme)
        score += platform_score * 0.1
        factors.append(("platform_optimization", platform_score))
        
        # Store analysis details
        meme.performance_metrics['virality_factors'] = factors
        meme.performance_metrics['virality_score'] = score
        
        return min(1.0, score)
    
    def _analyze_text_virality(self, text_content: List[str]) -> float:
        """Analyze text content for viral potential"""
        if not text_content:
            return 0.0
        
        combined_text = " ".join(text_content).lower()
        score = 0.0
        
        # Check for viral keywords
        for keyword, weight in self.viral_keywords.items():
            if keyword in combined_text:
                score += weight
        
        # Length optimization (memes should be concise)
        avg_length = sum(len(text) for text in text_content) / len(text_content)
        if 10 <= avg_length <= 50:  # Optimal length range
            score += 0.2
        elif avg_length > 100:  # Too long
            score -= 0.1
        
        # Check for relatability patterns
        relatability_patterns = ["when you", "me trying to", "that feeling when", "me:", "also me:"]
        for pattern in relatability_patterns:
            if pattern in combined_text:
                score += 0.3
                break
        
        # Check for humor indicators
        humor_indicators = ["literally", "actually", "basically", "apparently", "obviously"]
        humor_count = sum(1 for indicator in humor_indicators if indicator in combined_text)
        score += min(0.2, humor_count * 0.1)
        
        return min(1.0, score)
    
    def _analyze_topic_virality(self, topic: str, context: Dict[str, Any] = None) -> float:
        """Analyze topic viral potential"""
        score = 0.5  # Base score
        
        # Current events bonus
        current_event_keywords = ['breaking', 'news', 'trending', 'viral', 'controversy', 'scandal']
        topic_lower = topic.lower()
        
        for keyword in current_event_keywords:
            if keyword in topic_lower:
                score += 0.2
        
        # Technology/social media topics tend to be more viral
        tech_keywords = ['ai', 'tech', 'social media', 'internet', 'crypto', 'nft']
        for keyword in tech_keywords:
            if keyword in topic_lower:
                score += 0.15
        
        # Context-based scoring
        if context:
            trend_score = context.get('trend_score', 0.5)
            engagement_rate = context.get('engagement_rate', 0.5)
            score += (trend_score + engagement_rate) * 0.1
        
        return min(1.0, score)
    
    def _analyze_emotional_impact(self, text_content: List[str]) -> float:
        """Analyze emotional impact of meme content"""
        if not text_content or not self.classifier:
            return 0.5
        
        try:
            combined_text = " ".join(text_content)
            emotions = self.classifier(combined_text)[0]
            
            # High-engagement emotions
            viral_emotions = ['joy', 'anger', 'surprise']
            score = 0.0
            
            for emotion in emotions:
                if emotion['label'].lower() in viral_emotions:
                    score += emotion['score'] * 0.5
            
            return min(1.0, score)
            
        except Exception as e:
            logger.warning(f"Emotion analysis failed: {e}")
            return 0.5
    
    def _analyze_platform_fit(self, meme: GeneratedMeme) -> float:
        """Analyze how well the meme fits target platforms"""
        score = 0.0
        
        for platform in meme.target_platforms:
            platform_score = self._get_platform_compatibility(meme, platform)
            score += platform_score
        
        return score / len(meme.target_platforms) if meme.target_platforms else 0.5
    
    def _get_platform_compatibility(self, meme: GeneratedMeme, platform: Platform) -> float:
        """Get compatibility score for specific platform"""
        score = 0.5
        
        # Platform-specific optimizations
        if platform == Platform.TWITTER:
            # Twitter favors text-heavy, current events
            if len(meme.hashtags) > 0:
                score += 0.2
            if any(len(text) <= 280 for text in meme.text_content):
                score += 0.3
        
        elif platform == Platform.INSTAGRAM:
            # Instagram favors visual content
            if meme.template.format in [MemeFormat.DRAKE_POINTING, MemeFormat.DISTRACTED_BOYFRIEND]:
                score += 0.3
        
        elif platform == Platform.TIKTOK:
            # TikTok favors short, punchy content
            avg_length = sum(len(text) for text in meme.text_content) / len(meme.text_content)
            if avg_length <= 25:
                score += 0.4
        
        elif platform == Platform.REDDIT:
            # Reddit favors detailed, reference-heavy content
            if meme.template.format in [MemeFormat.EXPANDING_BRAIN, MemeFormat.CHANGE_MY_MIND]:
                score += 0.3
        
        return min(1.0, score)
    
    def get_virality_level(self, score: float) -> ViralityLevel:
        """Convert virality score to level"""
        if score >= 0.8:
            return ViralityLevel.VIRAL
        elif score >= 0.6:
            return ViralityLevel.HIGH
        elif score >= 0.4:
            return ViralityLevel.MEDIUM
        else:
            return ViralityLevel.LOW
    
    def suggest_improvements(self, meme: GeneratedMeme) -> List[str]:
        """Suggest improvements to increase viral potential"""
        suggestions = []
        
        # Analyze current score
        factors = meme.performance_metrics.get('virality_factors', [])
        
        for factor_name, score in factors:
            if score < 0.5:
                if factor_name == 'text_analysis':
                    suggestions.append("Add more relatable language like 'when you...' or 'me trying to...'")
                    suggestions.append("Include trending keywords or slang")
                elif factor_name == 'emotional_impact':
                    suggestions.append("Add more emotional language to increase engagement")
                elif factor_name == 'platform_optimization':
                    suggestions.append("Optimize text length for target platforms")
                    suggestions.append("Add relevant hashtags for discoverability")
        
        # General suggestions
        if meme.virality_score < 0.6:
            suggestions.append("Consider using a more popular meme template")
            suggestions.append("Make the content more relatable to your target audience")
            suggestions.append("Add current event references for timeliness")
        
        return suggestions[:5]  # Limit to top 5 suggestions
````

````python
class MemeGenerator:
    """Main meme generation orchestrator"""
    
    def __init__(self):
        self.trend_analyzer = TrendAnalyzer()
        self.text_generator = MemeTextGenerator()
        self.virality_predictor = ViralityPredictor()
        
        self.meme_templates = self._initialize_templates()
        self.generated_memes = []
        self.performance_history = {}
    
    def _initialize_templates(self) -> Dict[MemeFormat, MemeTemplate]:
        """Initialize meme templates"""
        templates = {}
        
        # Define text positions for different meme formats
        format_configs = {
            MemeFormat.DRAKE_POINTING: {
                "positions": [
                    {"x": 200, "y": 100, "width": 300, "height": 100},
                    {"x": 200, "y": 300, "width": 300, "height": 100}
                ],
                "popularity": 0.9,
                "context": ["comparison", "preference", "choice"]
            },
            MemeFormat.DISTRACTED_BOYFRIEND: {
                "positions": [
                    {"x": 50, "y": 50, "width": 150, "height": 50},  # boyfriend label
                    {"x": 350, "y": 50, "width": 150, "height": 50},  # new girl label
                    {"x": 200, "y": 300, "width": 150, "height": 50}   # girlfriend label
                ],
                "popularity": 0.8,
                "context": ["temptation", "distraction", "choice"]
            },
            MemeFormat.WOMAN_YELLING_CAT: {
                "positions": [
                    {"x": 50, "y": 300, "width": 200, "height": 100},
                    {"x": 350, "y": 300, "width": 200, "height": 100}
                ],
                "popularity": 0.7,
                "context": ["argument", "confusion", "misunderstanding"]
            },
            MemeFormat.EXPANDING_BRAIN: {
                "positions": [
                    {"x": 200, "y": 50, "width": 300, "height": 50},
                    {"x": 200, "y": 150, "width": 300, "height": 50},
                    {"x": 200, "y": 250, "width": 300, "height": 50},
                    {"x": 200, "y": 350, "width": 300, "height": 50}
                ],
                "popularity": 0.6,
                "context": ["intelligence", "progression", "enlightenment"]
            },
            MemeFormat.SURPRISED_PIKACHU: {
                "positions": [
                    {"x": 50, "y": 50, "width": 400, "height": 100},
                    {"x": 50, "y": 300, "width": 400, "height": 100}
                ],
                "popularity": 0.8,
                "context": ["surprise", "obvious", "consequence"]
            },
            MemeFormat.CHANGE_MY_MIND: {
                "positions": [
                    {"x": 100, "y": 200, "width": 300, "height": 100}
                ],
                "popularity": 0.5,
                "context": ["opinion", "controversial", "debate"]
            },
            MemeFormat.THIS_IS_FINE: {
                "positions": [
                    {"x": 50, "y": 50, "width": 200, "height": 50},
                    {"x": 300, "y": 300, "width": 150, "height": 50}
                ],
                "popularity": 0.7,
                "context": ["disaster", "denial", "calm"]
            },
            MemeFormat.STONKS: {
                "positions": [
                    {"x": 100, "y": 50, "width": 300, "height": 100},
                    {"x": 200, "y": 300, "width": 100, "height": 50}
                ],
                "popularity": 0.6,
                "context": ["finance", "investment", "success"]
            }
        }
        
        for meme_format, config in format_configs.items():
            template = MemeTemplate(
                template_id=f"template_{meme_format.value}",
                name=meme_format.value.replace('_', ' ').title(),
                format=meme_format,
                text_positions=config["positions"],
                popularity_score=config["popularity"],
                usage_context=config["context"]
            )
            templates[meme_format] = template
        
        return templates
    
    def generate_trending_meme(self, platform: Platform = Platform.TWITTER, 
                              topic_filter: str = None) -> GeneratedMeme:
        """Generate a meme based on current trending topics"""
        # Get trending topics
        trending_topics = self.trend_analyzer.get_trending_topics(platform)
        
        if topic_filter:
            trending_topics = [t for t in trending_topics if topic_filter.lower() in t.topic.lower()]
        
        if not trending_topics:
            raise ValueError("No suitable trending topics found")
        
        # Select best topic based on meme potential
        best_topic = max(trending_topics, key=lambda t: self.trend_analyzer.calculate_meme_opportunity_score(t))
        
        # Select appropriate meme format
        meme_format = self._select_meme_format(best_topic)
        
        # Generate meme
        return self._create_meme(best_topic, meme_format, [platform])
    
    def generate_market_event_meme(self, event: MarketEvent, 
                                  target_platforms: List[Platform]) -> GeneratedMeme:
        """Generate a meme based on a specific market event"""
        # Create synthetic trending topic from market event
        topic = TrendingTopic(
            topic=event.title,
            platform=target_platforms[0] if target_platforms else Platform.TWITTER,
            trend_score=event.impact_score,
            sentiment=event.sentiment,
            hashtags=[f"#{brand.replace(' ', '')}" for brand in event.related_brands[:3]],
            keywords=event.related_brands + event.related_stocks,
            discovered_at=event.occurred_at
        )
        
        # Select format based on event type
        event_format_mapping = {
            "earnings": MemeFormat.STONKS,
            "cultural": MemeFormat.DRAKE_POINTING,
            "market": MemeFormat.THIS_IS_FINE,
            "product_launch": MemeFormat.DISTRACTED_BOYFRIEND,
            "political": MemeFormat.WOMAN_YELLING_CAT
        }
        
        meme_format = event_format_mapping.get(event.event_type, MemeFormat.SURPRISED_PIKACHU)
        
        return self._create_meme(topic, meme_format, target_platforms, context=event.description)
    
    def generate_custom_meme(self, topic: str, meme_format: MemeFormat, 
                           target_platforms: List[Platform], 
                           custom_text: List[str] = None) -> GeneratedMeme:
        """Generate a custom meme with specified parameters"""
        # Create synthetic trending topic
        synthetic_topic = TrendingTopic(
            topic=topic,
            platform=target_platforms[0] if target_platforms else Platform.TWITTER,
            trend_score=0.5,
            sentiment="neutral",
            hashtags=[],
            keywords=topic.split(),
            discovered_at=datetime.now()
        )
        
        return self._create_meme(synthetic_topic, meme_format, target_platforms, custom_text=custom_text)
    
    def _create_meme(self, topic: TrendingTopic, meme_format: MemeFormat, 
                    target_platforms: List[Platform], context: str = "", 
                    custom_text: List[str] = None) -> GeneratedMeme:
        """Create a meme with all components"""
        # Get template
        template = self.meme_templates[meme_format]
        
        # Generate or use custom text
        if custom_text:
            text_content = custom_text
        else:
            text_content = self.text_generator.generate_meme_text(topic, meme_format, context)
        
        # Optimize for platforms
        optimized_text = []
        for platform in target_platforms:
            platform_text = self.text_generator.optimize_for_platform(text_content, platform)
            optimized_text.extend(platform_text)
        
        # Use the best optimized text
        final_text = optimized_text[:len(template.text_positions)] if optimized_text else text_content
        
        # Create meme object
        meme_id = f"meme_{int(datetime.now().timestamp())}_{hashlib.md5(topic.topic.encode()).hexdigest()[:8]}"
        
        meme = GeneratedMeme(
            meme_id=meme_id,
            template=template,
            text_content=final_text,
            topic=topic.topic,
            virality_score=0.0,  # Will be calculated
            sentiment=topic.sentiment,
            hashtags=topic.hashtags,
            target_platforms=target_platforms,
            brand_safe=self._check_brand_safety(final_text, topic)
        )
        
        # Predict virality
        virality_context = {
            'trend_score': topic.trend_score,
            'engagement_rate': 0.5  # Default
        }
        meme.virality_score = self.virality_predictor.predict_virality(meme, virality_context)
        
        # Store generated meme
        self.generated_memes.append(meme)
        
        return meme
    
    def _select_meme_format(self, topic: TrendingTopic) -> MemeFormat:
        """Select the most appropriate meme format for a topic"""
        # Topic-based format selection
        topic_lower = topic.topic.lower()
        
        if any(word in topic_lower for word in ['stock', 'finance', 'money', 'investment']):
            return MemeFormat.STONKS
        elif any(word in topic_lower for word in ['ai', 'technology', 'innovation']):
            return MemeFormat.EXPANDING_BRAIN
        elif any(word in topic_lower for word in ['drama', 'controversy', 'argument']):
            return MemeFormat.WOMAN_YELLING_CAT
        elif any(word in topic_lower for word in ['choice', 'comparison', 'versus']):
            return MemeFormat.DRAKE_POINTING
        elif any(word in topic_lower for word in ['disaster', 'crisis', 'problem']):
            return MemeFormat.THIS_IS_FINE
        elif topic.sentiment == 'negative':
            return MemeFormat.SURPRISED_PIKACHU
        else:
            # Default to popular formats
            popular_formats = [MemeFormat.DRAKE_POINTING, MemeFormat.DISTRACTED_BOYFRIEND, MemeFormat.SURPRISED_PIKACHU]
            return random.choice(popular_formats)
    
    def _check_brand_safety(self, text_content: List[str], topic: TrendingTopic) -> bool:
        """Check if content is brand-safe"""
        combined_text = " ".join(text_content).lower()
        
        # Basic profanity/controversial content filter
        unsafe_keywords = [
            'hate', 'violence', 'discrimination', 'offensive', 'inappropriate',
            'controversial', 'political', 'religion', 'nsfw'
        ]
        
        for keyword in unsafe_keywords:
            if keyword in combined_text:
                return False
        
        # Topic-based safety check
        if any(word in topic.topic.lower() for word in ['political', 'religion', 'controversial']):
            return False
        
        return True
    
    def get_meme_suggestions(self, target_audience: str = "general", 
                           platform: Platform = Platform.TWITTER) -> List[Dict[str, Any]]:
        """Get meme suggestions based on current trends"""
        trending_topics = self.trend_analyzer.get_trending_topics(platform)
        market_events = self.trend_analyzer.detect_market_events()
        
        suggestions = []
        
        # Suggest based on trending topics
        for topic in trending_topics[:3]:
            opportunity_score = self.trend_analyzer.calculate_meme_opportunity_score(topic)
            suggested_format = self._select_meme_format(topic)
            
            suggestion = {
                "type": "trending_topic",
                "topic": topic.topic,
                "suggested_format": suggested_format.value,
                "opportunity_score": opportunity_score,
                "platform": platform.value,
                "hashtags": topic.hashtags,
                "sentiment": topic.sentiment,
                "reasoning": f"High trending score ({topic.trend_score:.2f}) with {topic.sentiment} sentiment"
            }
            suggestions.append(suggestion)
        
        # Suggest based on market events
        for event in market_events[:2]:
            if event.meme_potential > 0.6:
                suggestion = {
                    "type": "market_event",
                    "topic": event.title,
                    "suggested_format": self._select_event_format(event).value,
                    "opportunity_score": event.meme_potential,
                    "platform": platform.value,
                    "event_type": event.event_type,
                    "impact_score": event.impact_score,
                    "reasoning": f"Market event with high meme potential ({event.meme_potential:.2f})"
                }
                suggestions.append(suggestion)
        
        # Sort by opportunity score
        suggestions.sort(key=lambda x: x['opportunity_score'], reverse=True)
        
        return suggestions
    
    def _select_event_format(self, event: MarketEvent) -> MemeFormat:
        """Select meme format based on market event"""
        event_formats = {
            "earnings": MemeFormat.STONKS,
            "product_launch": MemeFormat.DISTRACTED_BOYFRIEND,
            "cultural": MemeFormat.DRAKE_POINTING,
            "market": MemeFormat.THIS_IS_FINE,
            "political": MemeFormat.WOMAN_YELLING_CAT
        }
        
        return event_formats.get(event.event_type, MemeFormat.SURPRISED_PIKACHU)
    
    def analyze_meme_performance(self, meme_id: str) -> Dict[str, Any]:
        """Analyze performance of a generated meme"""
        meme = next((m for m in self.generated_memes if m.meme_id == meme_id), None)
        
        if not meme:
            return {"error": "Meme not found"}
        
        # Get virality prediction details
        virality_level = self.virality_predictor.get_virality_level(meme.virality_score)
        suggestions = self.virality_predictor.suggest_improvements(meme)
        
        return {
            "meme_id": meme_id,
            "topic": meme.topic,
            "format": meme.template.format.value,
            "virality_score": meme.virality_score,
            "virality_level": virality_level.value,
            "brand_safe": meme.brand_safe,
            "target_platforms": [p.value for p in meme.target_platforms],
            "performance_factors": meme.performance_metrics.get('virality_factors', []),
            "improvement_suggestions": suggestions,
            "generated_at": meme.generated_at.isoformat()
        }
````
